# Projeto LUMINA - Pipeline de Extração de Dados SISAGUA
Este notebook realiza a filtragem dos dados brutos do **SISAGUA (Ministério da Saúde)** para o município de **SANTO ANDRÉ - SP**.

Ele está estruturado em duas etapas principais:
1. **Controle Semestral:** Filtra parâmetros semestrais (agrotóxicos, metais pesados, radioatividade e substâncias orgânicas).
2. **Controle Mensal - Parâmetros Básicos:** Filtra os parâmetros de rotina mensal (**Fluoreto / Flúor**, Cloro Residual, Turbidez, pH e Coliformes) para a **Régua de Relacionamento de Saúde Preditiva**.

---
## 1. Controle Semestral (Agrotóxicos, Metais e Orgânicos)
- **Arquivo de Entrada:** `controle_semestral_2026.csv`
- **Arquivo Gerado:** `sisagua_santo_andre_2026.csv`
- **Identificador de Município:** 4ª coluna (índice `3` - `NO_MUNICIPIO`)

```python
import pandas as pd

# 1. Lê o arquivo de Controle Semestral em lotes (chunksize) com encoding latin-1
pedacos_semestral = pd.read_csv(
    'controle_semestral_2026.csv',
    chunksize=10000,
    sep=';',
    encoding='latin-1',
    low_memory=False
)

# 2. Filtra Santo André pela 4ª coluna (índice 3)
santo_andre_semestral = pd.concat(
    pedaco[pedaco.iloc[:, 3].astype(str).str.contains('SANTO ANDRE', na=False, case=False)]
    for pedaco in pedacos_semestral
)

# 3. Salva o arquivo filtrado de Controle Semestral
santo_andre_semestral.to_csv('C:\\Users\lucas\OneDrive\Área de Trabalho\Projeto LUMINA\Base de Dados\Arquivos SISAGUA\Tratado - Sto Andre\sisagua_semestral_santo_andre_2026.csv', index=False, sep=';', encoding='utf-8')
print(f"[Controle Semestral] Sucesso! Total de registros salvos: {len(santo_andre_semestral)}")
```

---
## 2. Controle Mensal - Parâmetros Básicos (Fluoreto / Flúor, Cloro, Turbidez, pH)
- **Arquivo de Entrada:** `controle_mensal_parametros_basicos_2026.csv`
- **Arquivo Gerado:** `sisagua_santo_andre_mensal_parametros_basicos_2026.csv`
- **Identificador de Município:** 5ª coluna (índice `4` - `Município`) ou 4ª coluna (índice `3` - `Código IBGE`)

```python
import pandas as pd

# 1. Lê o arquivo de Controle Mensal em lotes (chunksize) com encoding latin-1
pedacos_mensal = pd.read_csv(
    'C:\\Users\lucas\OneDrive\Área de Trabalho\Projeto LUMINA\Base de Dados\Arquivos SISAGUA\controle_mensal_parametros_basicos_2026.csv',
    chunksize=50000,
    sep=';',
    encoding='latin-1',
    low_memory=False
)

# 2. Filtra Santo André pela 5ª coluna (índice 4 - Município)
santo_andre_mensal = pd.concat(
    pedaco[pedaco.iloc[:, 4].astype(str).str.contains('SANTO ANDRE', na=False, case=False)]
    for pedaco in pedacos_mensal
)

# 3. Salva a base completa filtrada de Parâmetros Básicos de Santo André
santo_andre_mensal.to_csv('C:\\Users\lucas\OneDrive\Área de Trabalho\Projeto LUMINA\Base de Dados\Arquivos SISAGUA\Tratado - Sto Andre\sisagua_santo_andre_mensal_parametros_basicos_2026.csv', index=False, sep=';', encoding='utf-8')
print(f"[Controle Mensal] Sucesso! Total de registros salvos: {len(santo_andre_mensal)}")
```

---
## 3. Extração Específica de Fluoreto (Flúor) para a Régua de Relacionamento
Isola exclusivamente as medições do parâmetro **Fluoreto (mg/L)** em Santo André para alimentar os alertas preventivos do Lumina.

- **Arquivo Gerado:** `sisagua_santo_andre_fluoreto_2026.csv`
- **Coluna de Parâmetro:** 20ª coluna (índice `19` - `Parâmetro`)
- **Coluna de Métrica:** 21ª coluna (índice `20` - `Campo`: Percentil 95, Quantidade de Amostras, etc.)
- **Coluna de Concentração:** 22ª coluna (índice `21` - `Valor` em mg/L)

```python
# 1. Filtra apenas as linhas onde o Parâmetro (índice 19) é Fluoreto
coluna_parametro = santo_andre_mensal.columns[19]
fluor_df = santo_andre_mensal[santo_andre_mensal[coluna_parametro].astype(str).str.contains('Flu', case=False, na=False)]

# 2. Salva o CSV exclusivo de Fluoreto
fluor_df.to_csv('sisagua_santo_andre_fluoreto_2026.csv', index=False, sep=';', encoding='utf-8')
print(f"[Fluoreto] Sucesso! Total de registros de Fluoreto em Santo André: {len(fluor_df)}")

# 3. Exibe uma prévia das primeiras linhas
fluor_df[['Mês de referência', 'Nome da Forma de Abastecimento', 'Ponto de Monitoramento', 'Parâmetro', 'Campo', 'Valor']].head(15)
```
